**IMPORT PANDAS**

```python
import pandas as pd
```

**IMPORT EXCEL FILE THROUGH RELATIVE PATHING**

```python
file_path = "../data/processed/uk_rents_jun_2026.xlsx"
```

**CREATE DATAFRAME**

```python
df_jun_2026 = pd.read_excel(file_path, sheet_name='jun_26')

df_jun_2026.head(2)
```

**SLICE ONLY RENTAL PRICES, DO NOT INCLUDE INDICES NOR CHANGE RATES (WHETHER ANNUAL OR MONTHLY)**

```python
df_prices = df_jun_2026.loc[:, ["Time period", "Area code", "Area name", 
                                "Region or country name", "Rental price", "Rental price one bed",
                                "Rental price two bed", "Rental price three bed", "Rental price four or more bed",
                                "Rental price detached", "Rental price semidetached", "Rental price terraced",
                                "Rental price flat maisonette"]]

df_prices.head(2)
```

**NORTHERN IRELAND BOOLEAN MASK (In order to effectively verify that NI has not data)**

```python
df_prices[df_prices['Region or country name'] == 'Northern Ireland']
```

**SLICE NORTHERN IRELAND OFF, AND TEST IF THE SLICE WENT WELL WITH THE TAIL METHOD (THE LAST ENTRIES SHOULD SHOW "WALES")**

```python
df_prices = df_prices.loc[df_prices['Region or country name'] != 'Northern Ireland', :]

df_prices.tail(2)
```

**EXPORT**

```python
excel_path = "../data/processed/uk_rents_jun_2026.xlsx"

with pd.ExcelWriter(
    excel_path,
    engine='openpyxl',
    mode='a',
    if_sheet_exists='replace',
) as writer:

    df_prices.to_excel(writer, sheet_name='prices', index=False)
```
