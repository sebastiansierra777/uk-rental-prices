****IMPORT GODESS PANDAS****

```python
import pandas as pd
```

**IMPORT RAW ONS FILE THROUGH RELATIVE PATHING**

```python
file_path = "../data/raw/priceindexofprivaterentsukmonthlypricestatistics14.xlsx"
```

**SELECT CORRECT SHEET AND HEADER FOR DATAFRAME**

```python
df_rents = pd.read_excel(file_path, sheet_name='Table 1', header=2)

df_rents.head()
```

**SLICE JUN 2026 DATA**

```python
df_rents_jun_2026 = df_rents.loc[
    (df_rents["Time period"].dt.year == 2026) & (df_rents["Time period"].dt.month == 6)
]

df_rents_jun_2026.head(7)
```

**EXPORT JUN 2026 SLICE DATAFRAME TO A NEW EXCEL FILE INSIDE DATA/PROCESSED**

```python
excel_path = ("../data/processed/uk_rents_jun_2026.xlsx")

with pd.ExcelWriter(
    excel_path,
    engine='openpyxl',
    mode='w',
) as writer:

    df_rents_jun_2026.to_excel(writer, sheet_name='jun_26', index=False)
```
