**LIBRARY IMPORTS**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**EXCEL FILE IMPORT THROUGH RELATIVE PATHING**

```python
file_path = "../data/processed/uk_rents_jun_2026.xlsx"
```

**.head(17) IN ORDER TO VERIFY THAT THE FIRST 17 ROWS HAVE 'Region or country name' = '[z]' (BECAUSE THEY ARE THE AGGREGATE REGIONS THEMSELVES)**

```python
df_prices = pd.read_excel(file_path, sheet_name='prices')

df_prices.head(17)
```

**BOOLEAN MASK IN ORDER TO FURTHER VERIFY '[z]' SITUATION (THE NUMBER OF ENTRIES SHOULD BE EQUAL TO 14)**

```python
df_prices[df_prices['Region or country name'] == '[z]']
```

**BOOLEAN SLICE OF '[Z]' REGIONS**

```python
df_hist = df_prices.loc[df_prices['Region or country name'] != '[z]', :]

df_hist.head(3)
```

**GENERATE HISTOGRAM AND EXPORT**

```python
plt.figure(figsize=(12, 8))

sns.histplot(
    data=df_hist,
    x='Rental price',
    hue='Region or country name',
    bins=33,
    multiple='stack',
    kde=False,
    palette='Paired'
)

plt.savefig("../output/rental_prices_histogram.png", dpi=300, bbox_inches='tight')
```
